```python
import pandas as pd

roll_number = input("Enter your roll number: ")
last_two = roll_number[-2:]

categories = ["billing", "account", "general"]

fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },
    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },
    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },
    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]

new_entries = []

for digit in last_two:
    d = int(digit)
    category = categories[d % 3]

    if category == "billing":
        question = "how can i check my fee payment status"
        answer = "You can check your fee payment status from the payment section."
        keywords = "payment status fee"

    elif category == "account":
        question = "how do i update my registered mobile number"
        answer = "You can update your registered mobile number from account settings."
        keywords = "mobile number update"

    else:
        question = "where can i get general information"
        answer = "You can get general information from the college help desk."
        keywords = "information help general"

    new_entries.append({
        "question": question,
        "answer": answer,
        "keywords": keywords,
        "category": category
    })

df = pd.DataFrame(fixed_entries + new_entries)

print("\nFinal DataFrame:")
print(df)


def score_query(query, df):
    words = set(query.lower().split())
    results = []

    for _, row in df.iterrows():
        keywords = set(row["keywords"].lower().split())
        score = len(words & keywords)

        if score > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "score": score
            })

    return sorted(results, key=lambda x: x["score"], reverse=True)


query = input("\nEnter a query: ")
results = score_query(query, df)

print("\nMatching entries:")

if results:
    for r in results:
        print(r)
else:
    print("No matching FAQ found.")


def same_category(category_name, df):
    return df[df["category"] == category_name]


category = new_entries[0]["category"]

print("\nFAQs in", category, "category:")
print(same_category(category, df))


print("\nAdd a new keyword")
print("Question:", df.loc[0, "question"])

keyword = input("Enter new keyword: ")
df.loc[0, "keywords"] += " " + keyword

filename = roll_number + "_faq_data.csv"
df.to_csv(filename, index=False)

print("File saved as", filename)


print("\nFAQ count by category:")
print(df.groupby("category").size())


def score_query_with_tie(query, df):
    results = score_query(query, df)

    if not results:
        print("No matching entries found.")
        return

    highest = results[0]["score"]

    matches = [r for r in results if r["score"] == highest]

    if len(matches) > 1:
        print("\nTie found:")
        for r in matches:
            print(r)
    else:
        print("\nBest match:")
        print(matches[0])


print("\nTie example:")
score_query_with_tie("fee", df)

print("\nNo tie example:")
score_query_with_tie("password", df)
```
