```python
import numpy as np

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

for arr in [x, y]:
    values, counts = np.unique(arr, return_counts=True)
    max_count = counts.max()
    frequent = values[counts == max_count]

    print("\nArray:", arr)
    print("Most frequent value:", frequent)

    for value in frequent:
        print("Indices:", np.where(arr == value)[0])
```

```python
import numpy as np

nipun = np.linspace(10, 100, 25)

print("Array:")
print(nipun)

print("Dimensions:", nipun.ndim)
print("Shape:", nipun.shape)
print("Total elements:", nipun.size)
print("Data type:", nipun.dtype)
print("Total bytes:", nipun.nbytes)

transpose_nipun = nipun.reshape(25, 1)

print("Transpose using reshape:")
print(transpose_nipun)

print("Using T attribute:")
print(nipun.T)
```

```python
import numpy as np

ucs420_nipun = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Array:")
print(ucs420_nipun)

print("Mean:", np.mean(ucs420_nipun))
print("Median:", np.median(ucs420_nipun))
print("Maximum:", np.max(ucs420_nipun))
print("Minimum:", np.min(ucs420_nipun))
print("Unique elements:", np.unique(ucs420_nipun))

reshaped_ucs420_nipun = ucs420_nipun.reshape(4, 3)

print("\nReshaped array:")
print(reshaped_ucs420_nipun)

resized_ucs420_nipun = np.resize(ucs420_nipun, (2, 3))

print("\nResized array:")
print(resized_ucs420_nipun)
```
