```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
```

```python
temp_corrected = temperature + 2
print("Corrected temperatures:", temp_corrected)
```

```python
temp_fahrenheit = (9 / 5) * temperature + 32
print("Fahrenheit:", temp_fahrenheit)
```

```python
above_32 = temperature[temperature > 32]
print("Readings > 32°C:", above_32)
```

```python
count_above_32 = np.sum(temperature > 32)
print("Count of readings > 32°C:", count_above_32)
```

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])
```

```python
total_steps = np.sum(steps)
print("Total steps:", total_steps)
```

```python
mean_steps = np.mean(steps)
print("Mean steps:", mean_steps)
```

```python
max_steps = np.max(steps)
min_steps = np.min(steps)
print(f"Max steps: {max_steps}, Min steps: {min_steps}")
```

```python
daily_total = np.sum(steps, axis=0)
print("Total steps per day:", daily_total)
```

```python
user_total = np.sum(steps, axis=1)
print("Total steps per user:", user_total)
```

```python
max_flat_idx = np.argmax(steps)
max_pos = np.unravel_index(max_flat_idx, steps.shape)
print(f"Flat index: {max_flat_idx}, Matrix position (User, Day): {max_pos}")
```

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
y = np.array([40, 65, 30, 85])
```

```python
print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)
```

```python
X_T = X.T
print("X Transpose:\n", X_T)
```

```python
XTX = X_T @ X
print("X^T @ X:\n", XTX)
```

```python
XTX_inv = np.linalg.inv(XTX)
print("(X^T X)^(-1):\n", XTX_inv)
```

```python
beta = np.linalg.inv(X.T @ X) @ X.T @ y
print("Calculated Coefficients (Beta):", beta)
```

```python
new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta
print("Predicted Assistance Score:", predicted_score)
```
