## Question 1: Lists

**i. Build L from roll number digits × 10, print it**

```python
roll_digits = [int(d) for d in str(ROLL_NUMBER)]
L = [d * 10 for d in roll_digits]
print("L =", L)

# keep an untouched copy of the original 10-element L for use in Question 2
L_original = L.copy()
```

**ii. Add elements with append() and insert()**

```python
L.append(99)  # append() always adds the new element at the END of the list
print("After append(99):", L)

L.insert(3, 55)  # insert() adds the element at a chosen index (here, index 3), shifting the rest right
print("After insert(3, 55):", L)
```

**iii. Remove elements with remove() and pop()**

```python
L.remove(55)  # remove() deletes the first element that MATCHES the given value
print("After remove(55):", L)

popped = L.pop()  # pop() removes and returns the LAST element (or a given index)
print(f"After pop(): {L}  (popped value: {popped})")
```

**iv. Sort ascending then descending**

```python
L.sort()
print("Ascending:", L)

L.sort(reverse=True)
print("Descending:", L)
```

**v. Slicing — first three and last three elements**

```python
print("First three:", L[:3])
print("Last three:", L[-3:])
```

**vi. List comprehension — elements greater than the average**

```python
avg = sum(L) / len(L)
above_avg = [x for x in L if x > avg]
print(f"Average = {avg}")
print("Elements greater than average:", above_avg)
```

## Question 2: Tuples
Using the **original** 10-element L from Q1 (before append/insert/remove/pop), take the first 8 values.

```python
scores = tuple(L_original[:8])
print("scores =", scores)
```

**i. Highest score + its index, lowest score + its count**

```python
highest = max(scores)
highest_index = scores.index(highest)
lowest = min(scores)
lowest_count = scores.count(lowest)

print(f"Highest score: {highest} at index {highest_index}")
print(f"Lowest score: {lowest}, occurs {lowest_count} time(s)")
```

**ii. Reverse the tuple (as a new list)**

```python
reversed_scores = list(scores[::-1])
print("Reversed:", reversed_scores)

# Tuples are immutable, so there is no in-place .reverse() method like lists have.
# Reversing must build a brand-new sequence rather than modifying `scores` itself.
```

**iii. Search for a user-entered score**

```python
search_val = int(input("Enter a score to search for: "))

if search_val in scores:
    print(f"{search_val} found at index {scores.index(search_val)}")
else:
    print(f"{search_val} not present in scores")
```

**iv. Attempt to modify the tuple directly**

```python
try:
    scores[0] = 100
except TypeError as e:
    print(f"Error: {e}")
    # Tuples don't support item assignment because they're immutable —
    # lists allow scores[0] = 100 because lists ARE mutable.
```

**v. Unpack with \*: first, second, and the rest**

```python
first_score, second_score, *remaining_scores = scores
print("First:", first_score, "Second:", second_score, "Remaining:", remaining_scores)
```

## Question 3: Random numbers (seeded to your roll number)

```python
random.seed(int(ROLL_NUMBER))  # makes your list reproducible AND unique to you
```

**i. Generate 100 random numbers between 100 and 900 (inclusive)**

```python
random_list = [random.randint(100, 900) for _ in range(100)]
print(random_list)
```

**ii. Count odd numbers**

```python
odd_count = sum(1 for x in random_list if x % 2 != 0)
print("Odd numbers:", odd_count)
```

**iii. Count even numbers**

```python
even_count = sum(1 for x in random_list if x % 2 == 0)
print("Even numbers:", even_count)
```

**iv. Count and list prime numbers**

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

prime_list = [x for x in random_list if is_prime(x)]
print("Prime count:", len(prime_list))
print("Primes:", prime_list)
```

**v. Most frequent number**

```python
counts = Counter(random_list)
most_common_value, most_common_count = counts.most_common(1)[0]
print(f"Most frequent number: {most_common_value} (occurs {most_common_count} times)")
```

## Question 4: Sets
Using the original roll number digits (before ×10 from Q1).

```python
A = {d * 7 for d in roll_digits}
B = {d * 9 for d in roll_digits}
print("A =", A)
print("B =", B)
```

**vi. Union**

```python
print("Union:", A | B)
```

**vii. Intersection**

```python
print("Intersection:", A & B)
```

**viii. A − B and B − A separately**

```python
print("A - B:", A - B)
print("B - A:", B - A)

# difference() is DIRECTIONAL — A - B only keeps A's unique elements, B - A only keeps B's.
# symmetric_difference() (A ^ B) combines BOTH directions into a single set in one step.
```

**ix. Symmetric difference**

```python
print("Symmetric difference:", A ^ B)
```

**x. Subset / superset checks**

```python
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))
```

**xi. Discard a user-entered value from A**

```python
x_val = int(input("Enter a value to discard from set A: "))
A.discard(x_val)
print("A after discard:", A)

# discard() silently does nothing if the value isn't present, while remove() raises a KeyError.
# discard() is safer whenever you're not certain the value exists in the set.
```

## Question 5: Dictionary

```python
my_dict = {
    "name": NAME,
    "roll_no": ROLL_NUMBER,
    "branch": BRANCH,
    "age": AGE,
    "city": CITY
}
print(my_dict)
```

**i. Rename "city" → "location" (generic way, works for any dict)**

```python
my_dict["location"] = my_dict.pop("city")
print(my_dict)
```

**ii. Add "cgpa"**

```python
my_dict["cgpa"] = CGPA
print(my_dict)
```

**iii. Increase "age" by 1**

```python
my_dict["age"] += 1
print(my_dict)
```

**iv. Delete "branch" two ways, in two separate copies**

```python
dict_copy1 = my_dict.copy()
popped_branch = dict_copy1.pop("branch")
print("After pop():", dict_copy1, "| popped value:", popped_branch)

dict_copy2 = my_dict.copy()
del dict_copy2["branch"]
print("After del:", dict_copy2)

# pop() removes the key AND RETURNS its value, so you can capture what was deleted.
# del just deletes the key-value pair and returns nothing.
```

**v. Iterate with .items()**

```python
for key, value in my_dict.items():
    print(f"{key} \u2192 {value}")
```

**vi. Safely check for a missing key ("email")**

```python
if "email" in my_dict:
    print(my_dict["email"])
else:
    print("No email on file.")
```

**vii. Merge with a friend's dictionary**

```python
friend_dict = {
    "name": "Rahul Sharma",
    "roll_no": "9999999999",
    "branch": "ECE",
    "age": 21,
    "city": "Delhi"
}

merged = {**my_dict, **friend_dict}
print(merged)

# When keys overlap, whichever dictionary is listed LAST wins (friend_dict here),
# since its keys are unpacked after — and overwrite — my_dict's.
```

**viii. Dictionary comprehension — string-valued pairs only**

```python
string_only = {k: v for k, v in my_dict.items() if isinstance(v, str)}
print(string_only)
```
