## Question 6: Employees dataset

```python
emp_data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01", "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

emp_df = pd.DataFrame(emp_data)
emp_df.to_csv("employees.csv", index=False)

emp_df = pd.read_csv("employees.csv")
emp_df
```

**a) Shape of the DataFrame**

```python
emp_df.shape
```

**b) Summary — dtypes and non-null counts**

```python
emp_df.info()
```

**c) Descriptive statistics**

```python
emp_df.describe()
```

**d) First 5 rows and last 3 rows**

```python
print("First 5:")
display(emp_df.head())
print("Last 3:")
display(emp_df.tail(3))
```

**e) Key statistics**

```python
avg_salary = emp_df["Salary"].mean()
total_bonus = emp_df["Bonus"].sum()
youngest_age = emp_df["Age"].min()
highest_rating = emp_df["Rating"].max()

print(f"i.   Average salary: {avg_salary}")
print(f"ii.  Total bonus paid: {total_bonus}")
print(f"iii. Youngest employee's age: {youngest_age}")
print(f"iv.  Highest rating: {highest_rating}")
```

**f) Sort by Salary, descending**

```python
emp_df.sort_values("Salary", ascending=False)
```

**g) Performance category column**

```python
def categorize(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

emp_df["Performance_Category"] = emp_df["Rating"].apply(categorize)
emp_df[["Name", "Rating", "Performance_Category"]]
```

**h) Missing values**

```python
emp_df.isnull().sum()
```

**i) Rename Employee_ID → ID**

```python
emp_df = emp_df.rename(columns={"Employee_ID": "ID"})
emp_df.columns.tolist()
```

**j) Filter employees**

```python
print("i. More than 5 years of experience:")
display(emp_df[emp_df["Years_of_Experience"] > 5])

print("ii. Belong to IT department:")
display(emp_df[emp_df["Department"] == "IT"])
```

**k) Add Tax column (10% of Salary)**

```python
emp_df["Tax"] = emp_df["Salary"] * 0.10
emp_df[["Name", "Salary", "Tax"]]
```

**l) Save the modified DataFrame to a new CSV**

```python
emp_df.to_csv("employees_modified.csv", index=False)
print("Saved to employees_modified.csv")
```
